<h2>Part II: Getting Information from Dataset 1</h2>

```python
# Creating a violation_df that contains rows with actual violations
violation_df = df[df["ACTION"].isin([
    "Violations were cited in the following area(s).",
    "Establishment Closed by DOHMH. Violations were cited in the following area(s) and those requiring immediate action were addressed."
])]
violation_df = violation_df.drop(columns=["ACTION"]) # Dropping ACTION because it's already established that all of these restaurants had violations
violation_df
```

```python
# Finding violation counts per borough
violation_counts_per_borough = violation_df["BORO"].value_counts()

print("Violation counts per borough:")
print(violation_counts_per_borough)
```

```python
# Find the borough with the most violations and its count
borough_with_most_violations = violation_counts_per_borough.idxmax()
borough_with_most_violations_count = violation_counts_per_borough.max()

print(borough_with_most_violations, "is the borough with the most violations, with a total of", borough_with_most_violations_count, "violations.")
```

```python
# Finding the most common violation in all of NYC and its count
most_common_violation = violation_df["VIOLATION DESCRIPTION"].value_counts().idxmax()
most_common_violation_count = violation_df["VIOLATION DESCRIPTION"].value_counts().max()

print("The most common violation in all of NYC is:", most_common_violation)
print("This violation occured", most_common_violation_count, "times.")
```

```python
# Finding the most common violation per borough and its count
most_common_violation_per_borough = (violation_df.groupby("BORO")["VIOLATION DESCRIPTION"].apply(lambda x: x.value_counts().head(1))).reset_index()
most_common_violation_per_borough.columns = ["BORO", "VIOLATION DESCRIPTION", "COUNT"]

print("The most common violation per borough and its count:")
print(most_common_violation_per_borough)
```

```python
# Finding the borough and cuisine combination with the most violations
cuisine_borough_combo = violation_df.groupby(["BORO", "CUISINE DESCRIPTION"]).size().idxmax()
combo_violation_count = violation_df.groupby(['BORO', "CUISINE DESCRIPTION"]).size().max()

print("The borough and cuisine combination with the most violations is", cuisine_borough_combo[0],"and",cuisine_borough_combo[1],"cuisine.")
print("The number of violations for this combination is", combo_violation_count)
```

```python
# Finding the most common cuisine per borough and its count (not using violation_df)
most_common_cuisine_per_borough = (df.groupby("BORO")["CUISINE DESCRIPTION"].apply(lambda x: x.value_counts().head(1))).reset_index()
most_common_cuisine_per_borough.columns = ["BORO", "CUISINE DESCRIPTION", "COUNT"]

print("The most common cuisine per borough and its count:")
print(most_common_cuisine_per_borough)
```
